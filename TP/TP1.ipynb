```python
import shutil
import pkgutil

from pyparsing import line
def show_acceptable_modules () :
    line = '-' * 100
    print('{}\n{:^30}|{:^20}\n{}'. format(line , 'Module' , 'Location' , line) )
    for entry in pkgutil.iter_modules() :
        print('{:30}| {}'. format(entry [1] , entry [0]. path) )
```

```python
show_acceptable_modules()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/Life_Expectancy_Data.csv")
df = df.dropna()
df.info()
```

```python

df1 = df[(df.Country == "France") & (df.Year > 2010)]
print("df1: " , df1)
```

```python
df2 = df[(df.Country == "France") ] . Year
print("df2: " , df2)
```

```python
#df1 est un DataFrame qui contient les lignes du DataFrame df où la colonne "Country" est égale à "France" et où la colonne "Year" est supérieure à 2010. Cela signifie que df1 contient uniquement les données pour la France à partir de l'année 2011.
#df2 est une série (Series) qui contient uniquement les valeurs de la colonne "Year" du DataFrame df où la colonne "Country" est égale à "France". Cela signifie que df2 contient toutes les années pour lesquelles il y a des données pour la France, indépendamment de l'année.
```

```python
#question 2.2: A coder:
df_belgium = df[(df.Country == "Belgium") & (df.Year >= 2004) & (df.Year <= 2008)].sort_values(by="Year")["Life_expectancy"]

print ("Life expectancy in Belgium from 2004 to 2008: ", df_belgium.tolist())
    
```
