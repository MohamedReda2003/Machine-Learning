```python
#LE, ICoR = np.polyfit(df_study['Income_composition_of_resources'], df_study['Life_expectancy'], deg=1)
def moindre_carre(x,y):
    mean_x = np.mean(x)
    mean_y = np.mean(y)
    numerator = np.sum((x - mean_x) * (y - mean_y))
    denominator = np.sum((x - mean_x) ** 2)
    a = numerator / denominator
    b = mean_y - a * mean_x
    return a,b
LE, ICoR = moindre_carre(df_study['Income_composition_of_resources'], df_study['Life_expectancy'])

df_study['Life_expectancy_pred'] = LE * df_study['Income_composition_of_resources'] + ICoR
df_study['LE'] = LE
df_study['ICoR'] = ICoR
df_study
```

```python
# question 3.4:
def gradDescent(x, y, theta, alpha, iters):
    m = len(y)
    theta_values = []
    for _ in range(iters):
        prediction = np.dot(x, theta)
        error = prediction - y
        gradient = (1/m) * np.dot(x.T, error)
        theta -= alpha * gradient
        theta_values.append(theta.copy())
    #theta_sortie = theta
    return np.array(theta_values)
theta_0 = 0
theta_1 = 1
```
